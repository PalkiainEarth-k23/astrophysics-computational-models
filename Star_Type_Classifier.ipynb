```python
Star = input("What is the name of the star")
Temperature = int(input(f"What is the temeperature of {Star} in Kelvin: "))
if Temperature>= 30000:
    print(f"{Star} is a Type-O star")
elif 10000 <= Temperature <30000:
    print(f"{Star} is a Type- B star")
elif 7500<= Temperature < 10000:
    print(f"{Star} is an Type - A star")
elif 6000<= Temperature < 7500:
    print(f"{Star} is an Type - F star")
elif 5200<= Temperature < 6000:
    print(f"{Star} is an Type - G star")
elif 3500<= Temperature < 5200:
    print(f"{Star} is an Type - K star")
elif Temperature<=0:
    print(f"The lowest theoretical temperature is 0 kelvin, please enter a valid value for the temperature of the star")
else:
    print(f"{Star} is an Type - M star")
    
```
