```python
import math 
Name = input("What is the name of the planet whose escape velocity squared you want to find: ")
Mass = float(input(f"What is the mass of {Name}: "))
Radius = float(input(f"What is the radius of {Name}: "))
G = 0.00000000006674
Velocity_Squared = 2*(G * Mass) / Radius
Final_Escape_Velocity = math.sqrt(Velocity_Squared)
print(f"The final escape velocity of {Name} is : {Final_Escape_Velocity} ")
```
