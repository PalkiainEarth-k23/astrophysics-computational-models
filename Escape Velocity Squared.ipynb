```python
Name = input("What is the name of the planet whose escape velocity squared you want to find: ")
Mass = float(input(f"What is the mass of {Name}: "))
Radius = float(input(f"What is the radius of {Name}: "))
G = 0.00000000006674
Velocity_Squared = (G * Mass) / Radius
print(f"The square of the escape velocity of {Name} is : {Velocity_Squared} ")
```
