```python
Star = input("What is the name of the star")
mass = float(input("Enter {Star} mass (In solar masses): "))
Chandrasekhar_Limit = 1.44
if mass <= 0:
    print("Mass must be a positive number.")
elif mass > Chandrasekhar_Limit:
    print(f"Mass {mass} exceeds the Chandrasekhar limit! {Star} will collapse into a neutron star or black hole during the end of its lifetime.")
else:
    print(f"Mass {mass} is below the Chandrasekhar limit. The {Star} will become a white dwarf during the end of its lifetime .")
    
```
