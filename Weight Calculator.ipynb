```python
Mass = float(input("What is your mass(kg): "))
planet = input("Name of the planet : ")
Planet_Mass = float(input(f"What is the mass of the {planet} (in kg):"))
G = 0.00000000006674
R = int(input(f"What is the radius of {planet}"))
Your_new_weight = G *Planet_Mass*Mass/(R*R)
print(f"Your weight on {planet} is : {Your_new_weight; 2f} N")
```
