```python
c = 299792.458
H = 70
Galaxy = input("What is the name of the galaxy:")
Rest_Wavelength = float(input("What is the wavelength emmitted by the galaxy at rest state: "))
Observed_Wavelength = float(input("What is the apparent wavelength emmitted by the galaxy due to acceleration state: "))
Resession_V = ((Observed_Wavelength - Rest_Wavelength )/Rest_Wavelength)*c
Distance = Resession_V/H
if Resession_V>0:
    print(f"{Galaxy} is moving away from Earth at velocity {abs(Resession_V):.2f}km/s")
elif Resession_V<0:
    print(f"{Galaxy} is moving towards Earth due to local gravity with velocity {abs(Resession_V):.2f}km/s and is currently {abs(Distance):.2f} Megaparsecs away from Earth")
else:
    print(f"{Galaxy} is stationary with respect to Earth")

                        
```
