```python
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
import pandas as pd
```

```python
df = pd.read_csv("C:/Users/srini/Downloads/star_classification.csv")
df.head()
```

```python
X = df[['alpha', 'delta', 'u', 'g', 'r', 'i', 'z', 'redshift']]
dummies = pd.get_dummies(df[["class"]])
df = pd.concat([df, dummies], axis=1)
df.head()
```

```python
Y= df[['class_GALAXY', 'class_QSO', 'class_STAR']].astype(int)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2)
```

```python
model = keras.Sequential([
    keras.layers.Dense(128, activation='relu', input_shape=(8,)),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(32, activation='relu'),
    keras.layers.Dense(3, activation='softmax')
])
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(X_train, Y_train,epochs = 100)
```
